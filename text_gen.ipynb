```python
from transformers import pipeline, logging

logging.set_verbosity_error()

generator = pipeline(
    "text-generation",
    model="gpt2"
)

prompt = input("Enter a topic or starting sentence:\n\n")

result = generator(
    prompt,
    max_new_tokens=150,
    temperature=0.6,
    top_k=50,
    top_p=0.95,
    repetition_penalty=1.3,
    do_sample=True
)

generated_text = result[0]['generated_text']

print("\n===== USER PROMPT =====\n")
print(prompt)

print("\n **** GENERATED TEXT ****\n")
print(generated_text)

print("\nGenerated Word Count:", len(generated_text.split()))
```
